```python
fitted_estimates = pd.read_csv('/Users/proghani/Documents/personal/local_thesis_folder/fMRI/main/bids_fmri_data/sub-02/model_estimates_M37.csv')
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming fitted_estimates is your DataFrame
# If not, you'll need to create it or load it here

# Set the style
sns.set_style("whitegrid")

# Define custom colors
custom_colors = ['#e74c3c','#3498db']

# Adjust gridline transparency
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

# Set font sizes
SMALL_SIZE = 14
MEDIUM_SIZE = 16
LARGE_SIZE = 22

# Apply font sizes
plt.rc('font', size=SMALL_SIZE)
plt.rc('axes', titlesize=LARGE_SIZE)
plt.rc('axes', labelsize=MEDIUM_SIZE)
plt.rc('xtick', labelsize=SMALL_SIZE)
plt.rc('ytick', labelsize=SMALL_SIZE)
plt.rc('legend', fontsize=SMALL_SIZE)

# Create the plot
plt.figure(figsize=(12, 6))

# Plot the background effort
plt.plot(fitted_estimates['trialNinTask'], fitted_estimates['backgroundEffort'], 
         label='Background Effort', linestyle='--', color='#2c3e50', alpha=0.7)

# Overlay Block Type
for block_type, color in zip([99, 11], custom_colors):
    block_df = fitted_estimates[fitted_estimates['blockType'] == block_type]
    plt.scatter(block_df['trialNinTask'], block_df['backgroundEffort'], 
                color=color, label=f'Block Type {"Hard" if block_type == 99 else "Easy"}', 
                s=30, alpha=0.8)

# Customize the plot
plt.xlabel('Trial Number', fontsize=MEDIUM_SIZE)
plt.ylabel('Perceived Background Effort', fontsize=MEDIUM_SIZE)
plt.title('Evolution of Perceived Background Effort Over Trials - Low Learning Rate', 
          fontsize=LARGE_SIZE, pad=12)

# Adjust legend
plt.legend(title='', title_fontsize=SMALL_SIZE, loc='lower right', 
           bbox_to_anchor=(1, 0), frameon=True, edgecolor='black')

# Adjust layout and display the plot
plt.tight_layout()
plt.savefig('background_effort_evolution_low_lr.png', dpi=300, bbox_inches='tight')
plt.show()
```

## calcluating correlation between regressors of interest

```python
import statsmodels.api as sm

# Assuming your DataFrame is named 'fitted_estimates'
# Extract the relevant columns
X = fitted_estimates[['backgroundEffort']]
y = fitted_estimates['effortPE']

# Add a constant term for the intercept
X = sm.add_constant(X)

# Fit the regression model
model = sm.OLS(y, X).fit()

# Print the summary to get collinearity metrics
print(model.summary())
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Calculate the correlation
correlation = fitted_estimates['backgroundEffort'].corr(fitted_estimates['effortPE'])
print(f"Correlation between backgroundEffort and effortPE: {correlation}")

# Generate the scatterplot
plt.figure(figsize=(10, 6))
plt.scatter(fitted_estimates['backgroundEffort'], fitted_estimates['effortPE'], alpha=0.5)
plt.title(f'Scatterplot of Background Effort vs Effort Prediction Error\nCorrelation: {correlation:.2f}')
plt.xlabel('Background Effort')
plt.ylabel('Effort Prediction Error')
plt.grid(True)
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming fitted_estimates is your DataFrame
# If not, you'll need to create it or load it here

# Calculate the correlation
correlation = fitted_estimates['backgroundEffort'].corr(fitted_estimates['effortPE'])
# print(f"Correlation between backgroundEffort and effortPE: {correlation:.4f}")

# Set the style and color palette
sns.set_style("whitegrid")
sns.set_palette("deep")

# Adjust gridline transparency
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3  # Adjust this value to change transparency (0 to 1)

# Set font sizes
SMALL_SIZE = 10  # For axis labels and tick labels
MEDIUM_SIZE = 12  # For title

# Apply font sizes
plt.rc('font', size=SMALL_SIZE)  # controls default text sizes
plt.rc('axes', titlesize=MEDIUM_SIZE)  # fontsize of the axes title
plt.rc('axes', labelsize=SMALL_SIZE)  # fontsize of the x and y labels
plt.rc('xtick', labelsize=SMALL_SIZE)  # fontsize of the tick labels
plt.rc('ytick', labelsize=SMALL_SIZE)  # fontsize of the tick labels
plt.rc('legend', fontsize=SMALL_SIZE)  # legend fontsize

# Create the plot
plt.figure(figsize=(8, 6))
scatter_plot = sns.regplot(x='backgroundEffort', y='effortPE', data=fitted_estimates,
                           scatter_kws={'alpha': 0.5}, line_kws={'color': 'red'})

# Customize the plot
# scatter_plot.set_title(f'Background Effort vs Effort Prediction Error\nCorrelation: {correlation:.2f}', 
#                        fontsize=MEDIUM_SIZE)
scatter_plot.set_title(f'Correlation Analysis: Background Effort vs. Effort Prediction Error', 
                       fontsize=14)
scatter_plot.set_xlabel('Background Effort', fontsize=SMALL_SIZE)
scatter_plot.set_ylabel('Effort Prediction Error', fontsize=SMALL_SIZE)
scatter_plot.tick_params(labelsize=SMALL_SIZE)

# Add the correlation coefficient as text
plt.text(0.95, 0.95, f'Correlation: {correlation:.2f}', 
         transform=plt.gca().transAxes, fontsize=SMALL_SIZE, 
         verticalalignment='top', horizontalalignment='right')

# Adjust layout and display the plot
plt.tight_layout()

plt.savefig('regressors_correlation.png', dpi=300, bbox_inches='tight')

plt.show()
```
