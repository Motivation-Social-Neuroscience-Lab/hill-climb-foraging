```python
import nibabel as nib
from nilearn.image import resample_to_img
from nilearn.plotting import plot_stat_map
import matplotlib.pyplot as plt

# Load the functional image and the mask
func_img = nib.load('/Users/proghani/Documents/personal/local_thesis_folder/fMRI/main/bids_fmri_data/group_level_analysis/backgroundEff/final.nii')
mask_img = nib.load('/Users/proghani/Documents/personal/local_thesis_folder/fMRI/main/masks/neurovault/corrected_vmPFC_bin.nii')  # Update this path to your mask image

# Resample the mask to match the functional image
resampled_mask_img = resample_to_img(mask_img, func_img, interpolation='nearest')

# Get the data from the images
func_data = func_img.get_fdata()
mask_data = resampled_mask_img.get_fdata()

# Apply the mask
masked_data = func_data * mask_data

# Create a new NIfTI image for the masked data
masked_img = nib.Nifti1Image(masked_data, affine=func_img.affine, header=func_img.header)

# Visualize the functional image overlaid on the anatomical image with the mask applied
plot_stat_map(masked_img, bg_img=mni_img, threshold=15.3, vmax=50, cmap='hot', display_mode='ortho', cut_coords=(0, 0, 0))

plt.show()
```
